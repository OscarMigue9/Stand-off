```python
def transmittance_to_attenuation_db_m(array, path_length_m=1.0):
    wavelength, transmittance = array
    invL = 1.0 / max(path_length_m, 1e-300)
    transmittance_clipped = np.clip(transmittance, 1e-300, 1.0)
    attenuation = -(10.0 * invL) * np.log10(transmittance_clipped)
    return wavelength, attenuation
```

```python
def plot_attenuation_comparison(gt_atm, sim_atm, figsize=(14, 7), title="Atmospheric Attenuation Comparison"):
    from matplotlib.ticker import LogLocator
    
    # Función auxiliar para positivizar valores (evitar log de valores negativos/cero)
    def _positivize(arr):
        arr_pos = np.where(arr <= 0, 1e-300, arr)
        return arr_pos
    
    # Función auxiliar para calcular límites del eje logarítmico
    def _log_axis_limits(arr):
        arr_valid = arr[arr > 0]
        if arr_valid.size == 0:
            return (1e-10, 1e10)
        vmin, vmax = np.nanmin(arr_valid), np.nanmax(arr_valid)
        margin = 10.0
        return (vmin / margin, vmax * margin)
    
    # Preparar datos
    wavelength_gt, attenuation_gt = gt_atm
    wavelength_sim, attenuation_sim = sim_atm
    
    attenuation_gt_plot = _positivize(attenuation_gt)
    attenuation_sim_plot = _positivize(attenuation_sim)
    
    # Crear figura
    fig, ax = plt.subplots(figsize=figsize)
    
    # Plotear ambas curvas
    ax.semilogy(wavelength_gt, attenuation_gt_plot, lw=2.5, color="#d62728", 
                label="Ground Truth", alpha=0.8)
    ax.semilogy(wavelength_sim, attenuation_sim_plot, lw=2.0, color="#1f77b4", 
                label="Simulated", alpha=0.8, linestyle='--')
    
    # Configurar ejes y etiquetas
    ax.set_xlabel("Wavelength (µm)", fontsize=15)
    ax.set_ylabel("Attenuation (dB/m)", fontsize=15)
    ax.set_title(title, fontsize=18)
    
    # Determinar límites del eje Y basándose en ambos datasets
    all_data = np.concatenate([attenuation_gt_plot, attenuation_sim_plot])
    ax.set_ylim(*_log_axis_limits(all_data))
    
    # Configurar locator logarítmico
    ax.yaxis.set_major_locator(LogLocator(base=10.0))
    
    # Grids
    ax.grid(True, which='major', axis='both', color='#bbb', linestyle='-', 
            linewidth=0.8, alpha=0.5)
    ax.grid(True, which='minor', axis='both', color='#eee', linestyle=':', 
            linewidth=0.5, alpha=0.3)
    
    # Anotaciones de máximos
    max_idx_gt = np.nanargmax(attenuation_gt_plot)
    ax.annotate(f"GT Max: {attenuation_gt_plot[max_idx_gt]:.2e} dB/m",
                xy=(wavelength_gt[max_idx_gt], attenuation_gt_plot[max_idx_gt]),
                xytext=(wavelength_gt[max_idx_gt]+0.5, attenuation_gt_plot[max_idx_gt]*2),
                arrowprops=dict(arrowstyle="->", color="#d62728", lw=1.5), 
                fontsize=12, color="#d62728", bbox=dict(boxstyle="round,pad=0.3", 
                facecolor="white", edgecolor="#d62728", alpha=0.8))
    
    max_idx_sim = np.nanargmax(attenuation_sim_plot)
    ax.annotate(f"Sim Max: {attenuation_sim_plot[max_idx_sim]:.2e} dB/m",
                xy=(wavelength_sim[max_idx_sim], attenuation_sim_plot[max_idx_sim]),
                xytext=(wavelength_sim[max_idx_sim]+0.5, attenuation_sim_plot[max_idx_sim]/3),
                arrowprops=dict(arrowstyle="->", color="#1f77b4", lw=1.5), 
                fontsize=12, color="#1f77b4", bbox=dict(boxstyle="round,pad=0.3", 
                facecolor="white", edgecolor="#1f77b4", alpha=0.8))
    
    # Leyenda
    ax.legend(loc="upper right", frameon=True, fancybox=True, shadow=True, fontsize=13)
    
    plt.tight_layout()
    plt.show()
    
    return fig, ax
```
